```python
#Program 1, memfilter bilangan genap dan ganjil dari suatu list kemudian menjumlahkan berapa jumlah bilangan genap & ganjil nya

angka = (1, 2, 3, 4, 5, 6, 7, 8, 9)

genap = 0
ganjil = 0
i = 0

while i < len(angka):
    if angka[i] % 2 == 0:
        genap += 1
    else:
        ganjil += 1
    i += 1

print("list angka, ", angka)
print("Jumlah bilangan genap :", genap)
print("Jumlah bilangan ganjil :", ganjil)
```

```python
#Program 2, menampilkan nilai perkalian dari masukkan nilai

input_angka = int(input("Input sebuah angka: "))

for i in range(1, 11):
    hasil = input_angka * i
    print(f"{input_angka} x {i} = {hasil}")
```

```python
#Program 3, menampilkan urutan nilai dari fibonacci 

n = int(input("Masukkan jumlah suku Fibonacci: "))

a, b = 0, 1
count = 0

print("Deret Fibonacci:")
while count < n:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
print()
```

```python
#Program 4, perkalian nilai 1-100

for i in range(1, 11):
    for j in range(1, 11):
        print(i * j, end=" ")
    print()
```
